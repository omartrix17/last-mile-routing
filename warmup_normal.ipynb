With 10,000 draws from a normal(100, 20), how far off do you expect the sample mean to be from 100? You know the standard error formula from your stats classes, so use it to calculate a number.
Compute that expected error for 100 draws, 10,000 draws, and 1,000,000 draws. How much does each 100-fold increase in draws shrink the error? 
### Predictions

For a normal distribution with mean 100 and standard deviation 20, the standard error of the sample mean is SE = 20 / sqrt(n).

- For 100 draws, SE = 2.
- For 10,000 draws, SE = 0.2.
- For 1,000,000 draws, SE = 0.02.

With 10,000 draws, I predict that the sample mean will be close to 100, with a typical error of about 0.2. Each 100-fold increase in the number of draws reduces the standard error by a factor of 10.

```python
import numpy as np
import matplotlib.pyplot as plt
mu = 100
sigma = 20
sample_sizes = [100, 10000, 1000000]
kept = {}
for n in sample_sizes:
    se = sigma / np.sqrt(n)
    print (f"n={n: ,},Standard Error={se:.4f}")
rng = np.random.default_rng(17)
for n in sample_sizes:
    data = rng.normal(mu, sigma,size=n)
    sample_mean = np.mean(data)
    error = abs(sample_mean - mu)
    sd_error = abs(np.std(data) - sigma)
    print(f"Sample size:{n: ,}")
    print(f"Sample mean: {sample_mean: .4f}")
    print(f"Absolute error: {error: .4f}")
    print(f"standard deviation error:{sd_error: .4f}")
    if n == 10000:
        kept[n] = data
plt.hist(kept[10000], bins=50)
plt.axvline(mu, color ="red" , linestyle = "--")
plt.title("my first histogram, 10000 draws from Normal(100,20)")
plt.xlabel ("Value of each draw")
plt.ylabel("number of draws in each bin")
plt.show()
```
